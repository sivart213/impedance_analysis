```python
# Import Data
from pathlib import Path

from eis_analysis.system_utilities.file_io import load_file

# "D:\Online\ASU Dropbox\Jacob Clenney\Work Docs\Data\Analysis\IS\EVA\DC Analysis\rnd 2\converted_data.xlsx"
data_file = Path(
    r"D:\Online\ASU Dropbox\Jacob Clenney\Work Docs\Data\Analysis\IS\EVA\DC Analysis\rnd 2"
)

all_loaded_raw_data = {}
for file in ["converted_data.xlsx", "converted_100_and_301.xlsx"]:
    all_loaded_raw_data.update(load_file(data_file / file)[0])

print("Data loaded successfully.")
print(f"Total number of data points: {len(all_loaded_raw_data)}")
print(f"Data keys (ignoring runs): {set(k.split('_')[0] for k in all_loaded_raw_data.keys())}")
```
